## 01 · Research question

<div class="question-box">
<strong>Research question</strong><br>
Can the length of a numerical pattern predict whether it appears in the first 1,000,000 decimal digits of π?
</div>

<div class="insight-box">
<strong>Why this matters</strong><br>
In a finite sequence, longer exact patterns are harder to observe because the number of possible digit combinations grows rapidly with pattern length. This study tests whether pattern length alone contains enough information for prediction.
</div>

For a specific pattern of length <em>L</em>, a simple uniform-digit model gives the approximate expected number of occurrences:

<div class="equation-card">
  <div class="equation">E(L) ≈ N / 10<sup>L</sup></div>
  <div class="equation-note">N = number of digits searched · L = pattern length</div>
</div>

For <strong>N = 1,000,000</strong>:

<div class="expectation-grid">
  <div class="expectation-item"><strong>L = 4</strong>E(L) ≈ 100</div>
  <div class="expectation-item"><strong>L = 6</strong>E(L) ≈ 1</div>
  <div class="expectation-item"><strong>L = 8</strong>E(L) ≈ 0.01</div>
</div>

This provides a mathematical reason to expect pattern length to influence detectability.

The predictive relationship studied is:

**pattern length → found or not found**


## 02 · Data

The digits of π come from the [TidyTuesday dataset released on 24 March 2026](https://github.com/rfordatascience/tidytuesday/blob/main/data/2026/2026-03-24/readme.md).

The analysis uses the first **1,000,000 digits after the decimal point** and **150 predefined numerical patterns**:

- 50 patterns of length 4
- 50 patterns of length 6
- 50 patterns of length 8

For prediction:

- **X:** pattern length
- **y:** found (`1`) or not found (`0`)

<div class="stat-grid">
  <div class="stat-card"><span class="number">1M</span><span class="label">π digits searched</span></div>
  <div class="stat-card"><span class="number">150</span><span class="label">patterns</span></div>
  <div class="stat-card"><span class="number">3</span><span class="label">pattern lengths</span></div>
</div>


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.model_selection import train_test_split
from sklearn.dummy import DummyClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, ConfusionMatrixDisplay

# Load π digits
url = "https://raw.githubusercontent.com/rfordatascience/tidytuesday/main/data/2026/2026-03-24/pi_digits.csv"
df = pd.read_csv(url)
df = df.sort_values("digit_position").reset_index(drop=True)

# Remove the initial 3 and keep the first 1,000,000 decimal digits
pi_text = "".join(df["digit"].iloc[1:].astype(str))

print("Decimal digits available:", len(pi_text))
print("Opening digits:", pi_text[:30])


plt.rcParams.update({
    "figure.figsize": (7.6, 4.8),
    "figure.dpi": 130,
    "axes.facecolor": "#fbf8f1",
    "figure.facecolor": "#f4efe5",
    "axes.edgecolor": "#cfc5b5",
    "axes.labelcolor": "#1d2b34",
    "axes.titlecolor": "#1d2b34",
    "text.color": "#1d2b34",
    "xtick.color": "#66757d",
    "ytick.color": "#66757d",
    "grid.color": "#d8d1c5",
    "grid.alpha": 0.45,
    "font.family": "DejaVu Sans",
    "axes.titleweight": "bold",
})
```

## 03 · Pattern set

All patterns are defined **before** searching π to avoid selecting examples based on their outcome.

The set contains birthdays, repeated digits, counting sequences, palindromes, and arbitrary numbers. Pattern category is retained in the data but is not used as a predictor.


```python
groups = ["Birthdays", "Repeated", "Counting", "Palindromes", "Arbitrary"]
lengths = [4, 6, 8]

dates = [
    "24092001", "01011990", "14031985", "31122000", "29022004",
    "15061997", "07071977", "23111968", "05052005", "18081992"
]

palindromes = {
    4: ["1221", "2332", "3443", "4554", "5665", "6776", "7887", "8998", "9009", "0110"],
    6: ["123321", "234432", "345543", "456654", "567765", "678876", "789987", "890098", "901109", "012210"],
    8: ["12344321", "23455432", "34566543", "45677654", "56788765",
        "67899876", "78900987", "89011098", "90122109", "01233210"]
}

arbitrary = [
    "58302947", "92641703", "40719385", "85130629", "36902581",
    "70284916", "19476038", "63819520", "27580364", "04968257"
]

counting = "01234567890123456789"

rows = []

for length in lengths:
    for i in range(10):
        if length == 4:
            birthday = dates[i][:4]
        elif length == 6:
            birthday = dates[i][:4] + dates[i][-2:]
        else:
            birthday = dates[i]

        patterns = [
            birthday,
            str(i) * length,
            counting[i:i + length],
            palindromes[length][i],
            arbitrary[i][:length]
        ]

        for group, pattern in zip(groups, patterns):
            rows.append([group, length, pattern])

targets = pd.DataFrame(rows, columns=["category", "length", "pattern"])

print("Number of predefined patterns:", len(targets))
```

## 04 · Outcome

Each pattern is assigned a binary outcome:

- `1` if it appears at least once
- `0` if it does not appear

This produces the target variable used for classification.


```python
targets["found"] = targets["pattern"].apply(lambda pattern: int(pattern in pi_text))

summary = (
    targets.groupby("length")["found"]
    .agg(["count", "sum", "mean"])
    .rename(columns={"count": "patterns", "sum": "found", "mean": "found_rate"})
)

summary["found_percent"] = summary["found_rate"] * 100
display(summary[["patterns", "found", "found_percent"]].round(1))
```

```python
plot_data = summary.reset_index()

fig, ax = plt.subplots()

bars = ax.bar(
    plot_data["length"].astype(str),
    plot_data["found_percent"],
    color="#0f766e",
    width=0.58
)

ax.set_xlabel("Pattern length")
ax.set_ylabel("Patterns found (%)")
ax.set_title("Longer patterns are less likely to be found", pad=14)
ax.set_ylim(0, 105)
ax.grid(axis="y")
ax.spines[["top", "right"]].set_visible(False)

for bar, value in zip(bars, plot_data["found_percent"]):
    ax.text(
        bar.get_x() + bar.get_width()/2,
        value + 2,
        f"{value:.0f}%",
        ha="center",
        va="bottom",
        fontweight="bold"
    )

plt.show()
```

### Initial pattern

The observed match rate decreases sharply as pattern length increases:

- **4 digits:** 50/50 found
- **6 digits:** most found
- **8 digits:** almost none found

<div class="insight-box">
<strong>Key observation</strong><br>
Pattern length is strongly associated with whether a match is observed within a fixed search space.
</div>

## 05 · Predictive model

The task is binary classification: predict **found** or **not found** from pattern length.

A shallow **Decision Tree Classifier** estimates threshold-based relationships between length and outcome.

<div class="workflow">
  <div class="step"><b>1 · Data</b>150 patterns</div>
  <div class="step"><b>2 · Variables</b>X = length<br>y = found</div>
  <div class="step"><b>3 · Split</b>train / test</div>
  <div class="step"><b>4 · Fit</b>decision tree</div>
  <div class="step"><b>5 · Evaluate</b>baseline · accuracy · F1</div>
</div>

The data are divided into training and test sets. Model performance on the held-out test set is used to assess generalisation.


```python
# Feature and target
X = targets[["length"]]
y = targets["found"]

# Train/test split
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.25,
    random_state=42,
    stratify=y
)

# Create and fit the model
model = DecisionTreeClassifier(max_depth=2, random_state=42)
model.fit(X_train, y_train)

# Predict the test set
y_pred = model.predict(X_test)
```

### Learned decision rule

The fitted model separates the three observed pattern lengths using two thresholds.
The diagram below shows the learned rule without internal tree diagnostics.


```python
from IPython.display import display, HTML

length_check = pd.DataFrame({"length": [4, 6, 8]})
found_probs = model.predict_proba(length_check)[:, 1]
p4, p6, p8 = found_probs

tree_html = f"""
<div class="tree-wrap">

  <div class="tree-root">
    <div class="tree-node rule">
      <div class="tree-title">Is pattern length ≤ 7?</div>
      <div class="tree-sub">First learned threshold</div>
    </div>
  </div>

  <div class="tree-branch-labels">
    <div>Yes</div>
    <div>No</div>
  </div>

  <div class="tree-level">
    <div>
      <div class="tree-node rule">
        <div class="tree-title">Is pattern length ≤ 5?</div>
        <div class="tree-sub">Separates 4-digit from 6-digit patterns</div>
      </div>

      <div class="tree-leaves">
        <div class="tree-node found">
          <div class="tree-title">4 digits → Found</div>
          <div class="tree-sub">P(found) = {p4:.0%}</div>
        </div>
        <div class="tree-node found">
          <div class="tree-title">6 digits → Found</div>
          <div class="tree-sub">P(found) = {p6:.0%}</div>
        </div>
      </div>
    </div>

    <div>
      <div class="tree-node not-found">
        <div class="tree-title">8 digits → Not found</div>
        <div class="tree-sub">P(found) = {p8:.0%}</div>
      </div>
    </div>
  </div>

  <div class="model-note">
    Probabilities are estimated from training observations reaching each terminal node.
  </div>
</div>
"""

display(HTML(tree_html))
```

## 06 · Model assessment

The primary metric is **accuracy**:

$$
\text{Accuracy} =
\frac{\text{correct predictions}}
{\text{total predictions}}
$$

A **majority-class baseline** is also evaluated. It always predicts the most common class and provides a reference for whether the decision tree adds predictive value.

Additional metrics describe performance from different perspectives:

- **Precision:** proportion of predicted matches that are correct
- **Recall:** proportion of actual matches identified
- **F1 score:** balance between precision and recall


```python
# Majority-class baseline
baseline = DummyClassifier(strategy="most_frequent")
baseline.fit(X_train, y_train)
baseline_score = baseline.score(X_test, y_test)

# Decision-tree performance
train_score = model.score(X_train, y_train)
test_score = model.score(X_test, y_test)

results = pd.DataFrame({
    "Metric": [
        "Baseline accuracy",
        "Training accuracy",
        "Test accuracy",
        "Precision",
        "Recall",
        "F1 score"
    ],
    "Score": [
        baseline_score,
        train_score,
        test_score,
        precision_score(y_test, y_pred),
        recall_score(y_test, y_pred),
        f1_score(y_test, y_pred)
    ]
})

display(results.round(3))

from IPython.display import display, HTML

cards_html = (
    '<div class="stat-grid">'
    f'<div class="stat-card"><span class="number">{baseline_score:.0%}</span>'
    '<span class="label">baseline accuracy</span></div>'
    f'<div class="stat-card"><span class="number">{test_score:.0%}</span>'
    '<span class="label">decision-tree test accuracy</span></div>'
    f'<div class="stat-card"><span class="number">{f1_score(y_test, y_pred):.0%}</span>'
    '<span class="label">F1 score</span></div>'
    '</div>'
)

display(HTML(cards_html))
```

### Classification errors

The confusion matrix separates correct classifications from false positives and false negatives.

It shows the error structure behind the summary metrics.


```python
fig, ax = plt.subplots(figsize=(5.8, 5.0))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    display_labels=["Not found", "Found"],
    cmap="BuGn",
    colorbar=False,
    ax=ax
)

ax.set_title("Confusion matrix on the test set", pad=14)
plt.show()
```

## 07 · Interpretation

The theoretical relationship **E(L) ≈ N / 10<sup>L</sup>** predicts a rapid decline in expected occurrences as pattern length increases. The observed data follow the same general pattern.

The decision tree converts this relationship into simple length thresholds. Test accuracy is compared with a majority-class baseline to determine whether the predictor adds information beyond the most common outcome.

The confusion matrix shows where the simplified rule succeeds and where it fails.

<div class="insight-box">
<strong>Interpretation</strong><br>
Within a fixed one-million-digit search, pattern length contains useful predictive information because longer exact patterns have many more possible combinations and are therefore harder to encounter.
</div>

The model estimates **detectability from pattern length**; it does not test whether the digits of π are random.


## 08 · Example pattern lookup

A final example checks whether one numerical pattern appears in the first one million digits after the decimal point.

This lookup is illustrative and is not used for model training.


```python
my_pattern = "123321"  # Change this number

if not my_pattern.isdigit():
    print("Enter digits only.")
else:
    position = pi_text.find(my_pattern)

    if position == -1:
        print(my_pattern, "was not found in the first million decimal digits of π.")
    else:
        print(my_pattern, "was found.")
        print("First position after the decimal point:", position + 1)
```

## 09 · Conclusion

<div class="conclusion-box">
<strong>Finding</strong><br>
Pattern length is a useful predictor of whether a predefined number is observed within the first one million decimal digits of π.
</div>

The mathematical expectation

<div class="equation-card">
  <div class="equation">E(L) ≈ N / 10<sup>L</sup></div>
</div>

explains the central result: as pattern length increases, the number of possible exact digit strings grows exponentially, so individual patterns become harder to observe in a fixed search.

The machine-learning model captures this relationship using simple length thresholds and is evaluated against a majority-class baseline.

### Broader relevance

The same framework applies to other sequence-search problems in which **pattern size affects detectability**, including symbolic sequences, encoded data, and biological sequence analysis.

### Limitations

The 150 patterns are predefined and are not a random sample of all possible numerical strings. Results are therefore specific to this pattern set and search depth. The analysis does not test whether the digits of π are random.

### Future work

The analysis can be extended by:

- increasing the number and range of pattern lengths
- repeating the experiment at different search depths
- adding pattern structure as an additional predictor
- comparing multiple classification models
- applying the same framework to other numerical, symbolic, or biological sequences

These extensions would test whether the observed length effect remains stable under broader conditions.
